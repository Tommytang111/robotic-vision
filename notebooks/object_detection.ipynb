# Object detection + object tracking
### Uses fine-tuning of YOLOv11n + DeepSORT

```python
#Libraries
import cv2
from ultralytics import YOLO
#from resize_image import resize_image
from pathlib import Path
```

```python
#Load model
model = YOLO('yolov11n-fashionpedia.pt', task='detect')
```

```python
#Read video from webcam or video file and get video properties
video_source = 0
cap = cv2.VideoCapture(video_source)

if not cap.isOpened():
    print("Error: Could not open webcam/video.")
    exit()

#Video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

print(f"Video Properties: {width}x{height}, {fps} FPS")
```

```python
#Save video
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
output_filename = "output_detection.mp4"
out = cv2.VideoWriter(output_filename, fourcc, fps, (width, height))
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        print("End of video")
        break
    
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = model.predict(rgb_frame, conf=0.5)
    
    result = results[0]
    annotated_frame = result.plot() #Plots the results on the frame
    
    #Convert back to BGR for OpenCV
    bgr_annotated_frame = cv2.cvtColor(annotated_frame, cv2.COLOR_RGB2BGR)
    out.write(bgr_annotated_frame)
    
    cv2.imshow("YOLO Detection", bgr_annotated_frame))
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
    
cap.release()
out.release()
cv2.destroyAllWindows()
```
